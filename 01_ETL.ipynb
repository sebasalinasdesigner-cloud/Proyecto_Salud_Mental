# Fase 1: ETL - Preparación y Limpieza de Datos (CRISP-ML)

Este cuaderno aborda la primera fase del ciclo de vida del proyecto de machine learning: la Extracción, Transformación y Carga (ETL) de los datos.

### Objetivos:
1. Cargar el dataset original de adicción a redes sociales en estudiantes.
2. Analizar la estructura de los datos e identificar valores nulos o inconsistentes.
3. Limpiar el dataset (manejo de duplicados, corrección de tipos y rangos).
4. Exportar el dataset procesado para su posterior análisis (EDA) y modelado.

```python
import pandas as pd
import numpy as np

print("Librerías cargadas exitosamente.")
```

## 1. Carga de Datos

```python
# Cargar el dataset
df = pd.read_csv('Students Social Media Addiction.csv')
print(f"Dimensiones iniciales del dataset: {df.shape[0]} filas, {df.shape[1]} columnas.")
df.head()
```

## 2. Diagnóstico del Dataset (Calidad de Datos)

Vamos a explorar los tipos de datos de las columnas, la presencia de valores nulos o celdas vacías, y estadísticas descriptivas rápidas.

```python
# Información general de tipos y no-nulos
df.info()
```

```python
# Conteo de valores nulos por columna
print("Valores nulos por columna:")
print(df.isnull().sum())
```

```python
# Verificar duplicados
duplicates_count = df.duplicated().sum()
print(f"Cantidad de filas duplicadas completas: {duplicates_count}")
```

## 3. Limpieza y Transformación de Datos

```python
# Eliminar filas duplicadas si las hubiera
if duplicates_count > 0:
    df = df.drop_duplicates()
    print("Duplicados eliminados.")

# Validar rangos de valores para columnas críticas
print("Estadísticas de columnas numéricas clave:")
print(df[['Age', 'Avg_Daily_Usage_Hours', 'Sleep_Hours_Per_Night', 'Mental_Health_Score', 'Conflicts_Over_Social_Media', 'Addicted_Score']].describe())
```

```python
# Limpieza de cadenas de texto (quitar espacios en blanco al inicio/final)
text_cols = df.select_dtypes(include=['object']).columns
for col in text_cols:
    df[col] = df[col].astype(str).str.strip()

print("Limpieza de espacios en blanco en columnas de texto completada.")
```

## 4. Carga/Guardado del Dataset Limpio

```python
# Guardar el dataset limpio en un nuevo archivo CSV
output_path = 'Students_Social_Media_Addiction_Cleaned.csv'
df.to_csv(output_path, index=False)
print(f"Dataset limpio exportado exitosamente a '{output_path}'.")
```
