# Fase 2: EDA - Análisis Exploratorio de Datos (CRISP-ML)

En esta fase analizaremos el dataset limpio para descubrir patrones, tendencias y relaciones clave entre las variables y nuestro puntaje de adicción (`Addicted_Score`).

### Objetivos:
1. Comprender la distribución de las variables numéricas.
2. Evaluar el impacto de variables categóricas (género, plataforma más usada).
3. Analizar la correlación entre el uso de redes sociales, el sueño, la salud mental y los conflictos.
4. Generar visualizaciones claras e insights para justificar el uso de un modelo de Regresión Lineal.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración de estilos visuales elegantes
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['font.size'] = 12

print("Librerías de análisis y visualización importadas.")
```

## 1. Carga de Datos Limpios

```python
df = pd.read_csv('Students_Social_Media_Addiction_Cleaned.csv')
print(f"Dataset cargado con {df.shape[0]} registros.")
df.head()
```

## 2. Análisis Univariado

Exploraremos la distribución de la variable objetivo (`Addicted_Score`) y los principales predictores.

```python
# Distribución del Puntaje de Adicción
plt.figure(figsize=(8, 5))
sns.histplot(df['Addicted_Score'], bins=10, kde=True, color='purple')
plt.title('Distribución del Puntaje de Adicción a Redes Sociales')
plt.xlabel('Addicted Score (1 - 10)')
plt.ylabel('Frecuencia')
plt.show()
```

```python
# Distribución del uso de redes sociales diario vs horas de sueño
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

sns.histplot(df['Avg_Daily_Usage_Hours'], bins=15, kde=True, color='royalblue', ax=axes[0])
axes[0].set_title('Distribución del Uso Diario de Redes (Horas)')
axes[0].set_xlabel('Horas Diarias')
axes[0].set_ylabel('Frecuencia')

sns.histplot(df['Sleep_Hours_Per_Night'], bins=15, kde=True, color='teal', ax=axes[1])
axes[1].set_title('Distribución de Horas de Sueño por Noche')
axes[1].set_xlabel('Horas de Sueño')
axes[1].set_ylabel('Frecuencia')

plt.tight_layout()
plt.show()
```

## 3. Análisis Bivariado y Correlaciones

```python
# Matriz de correlación
numerical_cols = ['Age', 'Avg_Daily_Usage_Hours', 'Sleep_Hours_Per_Night', 'Mental_Health_Score', 'Conflicts_Over_Social_Media', 'Addicted_Score']
corr_matrix = df[numerical_cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".3f", linewidths=0.5)
plt.title('Matriz de Correlación entre Variables Numéricas')
plt.show()
```

```python
# Relación entre Horas de Uso y Puntaje de Adicción
plt.figure(figsize=(8, 5))
sns.regplot(data=df, x='Avg_Daily_Usage_Hours', y='Addicted_Score', color='crimson', scatter_kws={'alpha':0.6}, line_kws={'color':'darkblue'})
plt.title('Relación: Horas de Uso Diario vs. Puntaje de Adicción')
plt.xlabel('Horas de Uso Diario')
plt.ylabel('Puntaje de Adicción')
plt.show()
```

```python
# Relación entre Salud Mental y Puntaje de Adicción
plt.figure(figsize=(8, 5))
sns.regplot(data=df, x='Mental_Health_Score', y='Addicted_Score', color='forestgreen', scatter_kws={'alpha':0.6}, line_kws={'color':'darkred'})
plt.title('Relación: Puntaje de Salud Mental vs. Puntaje de Adicción')
plt.xlabel('Puntaje de Salud Mental')
plt.ylabel('Puntaje de Adicción')
plt.show()
```

## 4. Análisis por Categorías

```python
# Puntaje de adicción por Plataforma Más Usada
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='Most_Used_Platform', y='Addicted_Score', palette='Set2')
plt.title('Puntaje de Adicción por Plataforma Más Usada')
plt.xlabel('Plataforma')
plt.ylabel('Puntaje de Adicción')
plt.xticks(rotation=45)
plt.show()
```

### Conclusiones del EDA:
1. **Altas Correlaciones**: Existe una fuerte correlación lineal negativa entre `Mental_Health_Score` y `Addicted_Score` (-0.945), así como una fuerte correlación positiva entre `Conflicts_Over_Social_Media` y `Addicted_Score` (0.933).
2. **Uso Diario**: El uso promedio diario de redes tiene una correlación de 0.832 con la adicción, lo que valida la selección de este factor como predictor.
3. **Justificación del Modelo**: Las relaciones observadas tienen una fuerte tendencia lineal, lo que hace que la **Regresión Lineal** sea una opción de modelado inicial altamente adecuada y explicable.
